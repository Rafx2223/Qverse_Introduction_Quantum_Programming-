## 1. The Bell state

Start with `|00>`. Apply H to q0 and then CX with q0 as control and q1 as target:

$$
|\Phi^+\rangle = \frac{|00\rangle+|11\rangle}{\sqrt2}.
$$

Each qubit individually is unpredictable in the Z basis, but their outcomes are perfectly correlated.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, partial_trace
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import plot_histogram

bell = QuantumCircuit(2)
bell.h(0)
bell.cx(0, 1)

sv = Statevector.from_instruction(bell)
print(sv)
print(sv.probabilities_dict())
bell.draw("mpl")
```

```python
measured = bell.copy()
measured.measure_all()
counts = StatevectorSampler(seed=2).run([measured], shots=1000).result()[0].data.meas.get_counts()
print(counts)
plot_histogram(counts)
```

## 2. Each subsystem looks mixed

```python
rho_q0 = partial_trace(sv, [1])
rho_q1 = partial_trace(sv, [0])

print("Reduced state q0:\n", rho_q0.data)
print("Reduced state q1:\n", rho_q1.data)
```

The diagonal values `0.5, 0.5` mean either local Z measurement is random. Yet the **joint** state is not simply "q0 random and q1 random independently"; the outcomes are correlated.

## 3. Generate all four Bell states

```python
def bell_state(label):
    qc = QuantumCircuit(2)
    qc.h(0)
    qc.cx(0, 1)
    if label == "Phi+":
        pass
    elif label == "Phi-":
        qc.z(0)
    elif label == "Psi+":
        qc.x(1)
    elif label == "Psi-":
        qc.x(1)
        qc.z(0)
    else:
        raise ValueError("Choose Phi+, Phi-, Psi+, or Psi-")
    return qc

for label in ["Phi+", "Phi-", "Psi+", "Psi-"]:
    print(label, Statevector.from_instruction(bell_state(label)).data)
```

## 4. Measure Bell states in another basis

```python
for label in ["Phi+", "Phi-", "Psi+", "Psi-"]:
    qc = bell_state(label)
    # H before Z measurement is equivalent to measurement in the X basis.
    qc.h([0, 1])
    qc.measure_all()
    counts = StatevectorSampler(seed=11).run([qc], shots=500).result()[0].data.meas.get_counts()
    print(label, counts)
```
