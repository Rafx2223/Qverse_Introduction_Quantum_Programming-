## 1. Rotation gates

Rotation gates let us continuously tune a qubit state. A particularly intuitive example is

$$
R_y(\theta)|0\rangle =
\cos(\theta/2)|0\rangle+\sin(\theta/2)|1\rangle.
$$

Therefore $P(1)=\sin^2(\theta/2)$.

```python
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator

angles = np.linspace(0, 2*np.pi, 101)
p1 = []

for theta in angles:
    qc = QuantumCircuit(1)
    qc.ry(theta, 0)
    p1.append(Statevector.from_instruction(qc).probabilities()[1])

plt.plot(angles, p1)
plt.xlabel("θ")
plt.ylabel("P(1)")
plt.title("Ry rotation")
plt.show()
```

## 2. Special angles

```python
for theta in [0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi]:
    qc = QuantumCircuit(1)
    qc.ry(theta, 0)
    print(theta, Statevector.from_instruction(qc).probabilities_dict())
```

## 3. Controlled-X and the no-cloning warning

```python
qc = QuantumCircuit(2)
qc.x(0)       # known basis value
qc.cx(0, 1)   # copies the basis value 1
print(Statevector.from_instruction(qc).probabilities_dict())
qc.draw("mpl")
```

CX can copy an **already known computational-basis value**. It cannot clone an arbitrary unknown state.

If the control is in `|+>`, CX creates entanglement rather than two independent copies of `|+>`.

```python
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
print(Statevector.from_instruction(qc))
```

## 4. SWAP from three CX gates

```python
swap_from_cx = QuantumCircuit(2)
swap_from_cx.cx(0, 1)
swap_from_cx.cx(1, 0)
swap_from_cx.cx(0, 1)

built_in_swap = QuantumCircuit(2)
built_in_swap.swap(0, 1)

print("Equivalent:", Operator(swap_from_cx).equiv(Operator(built_in_swap)))
swap_from_cx.draw("mpl")
```

## 5. Circuit depth and operation counts

```python
print("Depth:", swap_from_cx.depth())
print("Operations:", swap_from_cx.count_ops())
```
