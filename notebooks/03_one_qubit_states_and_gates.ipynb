## 1. Four states to know

$$
|0\rangle=\begin{bmatrix}1\\0\end{bmatrix},\quad
|1\rangle=\begin{bmatrix}0\\1\end{bmatrix}
$$

$$
|+\rangle=\frac{|0\rangle+|1\rangle}{\sqrt2},\quad
|-\rangle=\frac{|0\rangle-|1\rangle}{\sqrt2}.
$$

`H|0> = |+>` and `H|1> = |->`.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import plot_histogram

states = {}

qc = QuantumCircuit(1)
states["|0>"] = Statevector.from_instruction(qc)

qc = QuantumCircuit(1); qc.x(0)
states["|1>"] = Statevector.from_instruction(qc)

qc = QuantumCircuit(1); qc.h(0)
states["|+>"] = Statevector.from_instruction(qc)

qc = QuantumCircuit(1); qc.x(0); qc.h(0)
states["|->"] = Statevector.from_instruction(qc)

for name, sv in states.items():
    print(name, sv.data)
```

## 2. Sampling `|+>`
Predict: will 1,000 shots give exactly 500 zeros and 500 ones?

```python
qc = QuantumCircuit(1)
qc.h(0)
qc.measure_all()

counts = StatevectorSampler(seed=21).run([qc], shots=1000).result()[0].data.meas.get_counts()
print(counts)
plot_histogram(counts)
```

## 3. The Z gate changes phase

`Z|0> = |0>` and `Z|1> = -|1>`. Therefore `Z|+> = |->`.

Notice the subtlety: `|+>` and `|->` have the **same Z-basis probabilities** even though they are different quantum states.

```python
qc_plus = QuantumCircuit(1)
qc_plus.h(0)

qc_minus = QuantumCircuit(1)
qc_minus.h(0)
qc_minus.z(0)

print("|+> :", Statevector.from_instruction(qc_plus).data)
print("Z|+>:", Statevector.from_instruction(qc_minus).data)
```

## 4. H is its own inverse

```python
for initial_one in (False, True):
    qc = QuantumCircuit(1)
    if initial_one:
        qc.x(0)
    qc.h(0)
    qc.h(0)
    print("Initial", int(initial_one), "->", Statevector.from_instruction(qc).probabilities_dict())
```

## 5. Global phase check

```python
import numpy as np
a = Statevector([1/np.sqrt(2), 1/np.sqrt(2)])
b = Statevector([1j/np.sqrt(2), 1j/np.sqrt(2)])

print("Same measurement probabilities?", np.allclose(a.probabilities(), b.probabilities()))
print("Equivalent up to global phase?", a.equiv(b))
```
