## 1. Interference is the algorithmic resource

Quantum gates transform **amplitudes**, including their signs and phases. When different computational paths recombine, amplitudes can add or cancel.

The sequence `H → Z → H` is a compact demonstration:
- first H creates two paths;
- Z changes the relative sign of one path;
- the second H recombines them;
- cancellation and reinforcement produce a deterministic output.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.primitives import StatevectorSampler

def sample_sequence(gates, shots=500):
    qc = QuantumCircuit(1)
    for gate in gates:
        getattr(qc, gate)(0)
    state_before_measurement = Statevector.from_instruction(qc)
    qc.measure_all()
    counts = StatevectorSampler(seed=5).run([qc], shots=shots).result()[0].data.meas.get_counts()
    return state_before_measurement, counts, qc

for seq in [("h","h"), ("h","z","h"), ("h","x","h")]:
    sv, counts, _ = sample_sequence(seq)
    print(seq, "state =", sv.data, "counts =", counts)
```

## 2. Verify the identity H Z H = X

```python
from qiskit.quantum_info import Operator
import numpy as np

qc_hzh = QuantumCircuit(1)
qc_hzh.h(0); qc_hzh.z(0); qc_hzh.h(0)

qc_x = QuantumCircuit(1)
qc_x.x(0)

print(np.allclose(Operator(qc_hzh).data, Operator(qc_x).data))
```

## 3. S and T gates

```python
for name in ["s", "t"]:
    qc = QuantumCircuit(1)
    qc.h(0)
    getattr(qc, name)(0)
    sv = Statevector.from_instruction(qc)
    print(name.upper(), "on |+>:", sv.data)
```

`S` adds a phase of $i=e^{i\pi/2}$ to the `|1>` amplitude.  
`T` adds a phase of $e^{i\pi/4}$.

If you immediately measure in Z, both still look 50/50. Add a suitable basis-changing operation and phase information can affect probabilities.

## 4. Interference scan with Rz

```python
import numpy as np
import matplotlib.pyplot as plt

angles = np.linspace(0, 2*np.pi, 100)
p1 = []

for theta in angles:
    qc = QuantumCircuit(1)
    qc.h(0)
    qc.rz(theta, 0)
    qc.h(0)
    sv = Statevector.from_instruction(qc)
    p1.append(sv.probabilities()[1])

plt.plot(angles, p1)
plt.xlabel("Rz angle θ")
plt.ylabel("P(1) after H-Rz(θ)-H")
plt.title("Phase converted into population")
plt.show()
```
