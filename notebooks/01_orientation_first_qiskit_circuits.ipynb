## 1. The central idea

A classical bit is stored as a definite `0` or `1`. A qubit is described by a quantum state. Later we will write a one-qubit state as

$$
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle.
$$

Do **not** interpret this as "the computer stores both answers and we can read both." Measurement returns a classical outcome, and quantum algorithms are designed so that useful outcomes become more probable through interference.

### Circuit vocabulary
- **Qubit:** quantum information carrier.
- **Gate:** controlled transformation of a quantum state.
- **Circuit:** ordered sequence of gates and measurements.
- **Measurement:** converts quantum information into a classical outcome.
- **Shot:** one execution/sample of a measured circuit.
- **Simulator:** classical software that models a quantum circuit.
- **QPU:** physical quantum processor.

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import plot_histogram
```

## 2. Empty circuit: what should happen?
**Prediction:** If a new qubit begins in `|0>` and we immediately measure it, what outcome should we obtain?

```python
qc0 = QuantumCircuit(1)
qc0.measure_all()
qc0.draw("mpl")
```

```python
sampler = StatevectorSampler(seed=7)
result = sampler.run([qc0], shots=200).result()
counts0 = result[0].data.meas.get_counts()
print(counts0)
plot_histogram(counts0)
```

## 3. The X gate
The Pauli-X gate swaps `|0>` and `|1>`. On computational-basis states it behaves like a classical NOT gate.

```python
qc1 = QuantumCircuit(1)
qc1.x(0)
qc1.measure_all()

counts1 = StatevectorSampler(seed=7).run([qc1], shots=200).result()[0].data.meas.get_counts()
print(counts1)
qc1.draw("mpl")
```

### Experiment: X repeated several times
Before running: predict the output after 0, 1, 2, 3, and 4 X gates.

```python
for repetitions in range(5):
    qc = QuantumCircuit(1)
    for _ in range(repetitions):
        qc.x(0)
    qc.measure_all()
    counts = StatevectorSampler(seed=12).run([qc], shots=50).result()[0].data.meas.get_counts()
    print(f"{repetitions} X gate(s): {counts}")
```

## 4. First reusable function

```python
def make_bit(value: int):
    if value not in (0, 1):
        raise ValueError("value must be 0 or 1")
    qc = QuantumCircuit(1)
    if value == 1:
        qc.x(0)
    qc.measure_all()
    return qc

make_bit(1).draw("mpl")
```
