## 1. Complex amplitudes

Quantum states use complex numbers. For introductory programming, the important operations are:
- complex conjugate;
- magnitude `abs(z)`;
- squared magnitude $|z|^2$;
- phase.

A valid one-qubit pure state can be represented by a two-component vector

$$
|\psi\rangle =
\begin{bmatrix}\alpha\\\beta\end{bmatrix},
\qquad |\alpha|^2+|\beta|^2=1.
$$

Measurement in the computational basis gives `0` with probability $|\alpha|^2$ and `1` with probability $|\beta|^2$.

```python
import numpy as np

z = 1 + 1j
print("z =", z)
print("|z| =", abs(z))
print("|z|^2 =", abs(z)**2)
print("phase =", np.angle(z), "radians")
```

## 2. Normalizing a candidate state

```python
psi = np.array([1, 1j], dtype=complex)
print("Before:", psi)
print("Norm:", np.linalg.norm(psi))

psi = psi / np.linalg.norm(psi)
probabilities = np.abs(psi)**2

print("Normalized:", psi)
print("Probabilities:", probabilities)
print("Sum:", probabilities.sum())
```

## 3. Gates as matrices

The X and H matrices are

$$
X=\begin{bmatrix}0&1\\1&0\end{bmatrix},
\qquad
H=\frac{1}{\sqrt2}\begin{bmatrix}1&1\\1&-1\end{bmatrix}.
$$

Matrix-vector multiplication tells us the new amplitudes.

```python
zero = np.array([1, 0], dtype=complex)
one  = np.array([0, 1], dtype=complex)

X = np.array([[0, 1], [1, 0]], dtype=complex)
H = (1/np.sqrt(2)) * np.array([[1, 1], [1, -1]], dtype=complex)

print("X|0> =", X @ zero)
print("H|0> =", H @ zero)
print("H|1> =", H @ one)
```

## 4. Let Qiskit represent the same state

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)
qc.h(0)
sv = Statevector.from_instruction(qc)

print("Statevector:", sv.data)
print("Probabilities:", sv.probabilities_dict())
```

## 5. A normalization helper

```python
def is_normalized(vector, atol=1e-10):
    vector = np.asarray(vector, dtype=complex)
    return np.isclose(np.sum(np.abs(vector)**2), 1.0, atol=atol)

print(is_normalized([1, 0]))
print(is_normalized([1, 1]))
print(is_normalized([1/np.sqrt(2), 1j/np.sqrt(2)]))
```
