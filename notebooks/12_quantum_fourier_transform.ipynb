## 1. Why Fourier ideas appear

The classical Fourier transform converts information between representations such as time and frequency. The Quantum Fourier Transform (QFT) is a unitary transformation on amplitudes:

$$
|j\rangle \mapsto \frac{1}{\sqrt{2^n}}
\sum_{k=0}^{2^n-1} e^{2\pi i jk/2^n}|k\rangle.
$$

For this introductory course, focus on:
- QFT encodes computational-basis information into a structured phase pattern;
- it can be implemented with Hadamards and controlled phase rotations;
- its inverse converts that structure back.

```python
from qiskit import QuantumCircuit
from qiskit.circuit.library import QFTGate
from qiskit.quantum_info import Statevector, Operator
```

## 2. Apply QFT to a basis state

```python
n = 3
qc = QuantumCircuit(n)
qc.x(0)  # prepare displayed basis state 001
before = Statevector.from_instruction(qc)

qft = QFTGate(n)
qc.append(qft, range(n))
after = Statevector.from_instruction(qc)

print("Before:", before.probabilities_dict())
print("After amplitudes:")
print(after.data)
qc.draw("mpl")
```

## 3. QFT followed by inverse QFT

```python
def qft_roundtrip(bitstring):
    n = len(bitstring)
    qc = QuantumCircuit(n)

    for q, bit in enumerate(reversed(bitstring)):
        if bit == "1":
            qc.x(q)

    qc.append(QFTGate(n), range(n))
    qc.append(QFTGate(n).inverse(), range(n))
    return qc

for bits in ["000", "001", "010", "101", "111"]:
    sv = Statevector.from_instruction(qft_roundtrip(bits))
    print(bits, sv.probabilities_dict())
```

## 4. Inspect the gate decomposition

```python
decomposed = QuantumCircuit(3)
decomposed.append(QFTGate(3), range(3))
print(decomposed.decompose(reps=2))
decomposed.decompose(reps=2).draw("mpl")
```

## 5. Scaling of decomposed circuit depth

```python
sizes = range(2, 9)
for n in sizes:
    qc = QuantumCircuit(n)
    qc.append(QFTGate(n), range(n))
    d = qc.decompose(reps=2)
    print(f"{n} qubits: depth={d.depth()}, ops={dict(d.count_ops())}")
```
