## 1. Two qubits

Two qubits have four computational-basis states:
$$
|00\rangle,\ |01\rangle,\ |10\rangle,\ |11\rangle.
$$

An arbitrary two-qubit pure state needs four amplitudes. In general, `n` qubits require $2^n$ amplitudes.

**Important Qiskit convention:** when bitstrings are printed, the highest-index classical bit appears on the left. We will test this rather than memorize it abstractly.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.primitives import StatevectorSampler

def show_state(qc, label):
    sv = Statevector.from_instruction(qc)
    print(label, sv.probabilities_dict())

for q0, q1 in [(0,0), (1,0), (0,1), (1,1)]:
    qc = QuantumCircuit(2)
    if q0: qc.x(0)
    if q1: qc.x(1)
    show_state(qc, f"q0={q0}, q1={q1}")
```

## 2. Product superposition: `|+>` on q0 and `|1>` on q1

```python
qc = QuantumCircuit(2)
qc.h(0)
qc.x(1)
sv = Statevector.from_instruction(qc)

print(sv.data)
print(sv.probabilities_dict())
qc.draw("mpl")
```

## 3. Measure and inspect ordering

```python
measured = qc.copy()
measured.measure_all()
counts = StatevectorSampler(seed=4).run([measured], shots=500).result()[0].data.meas.get_counts()
print(counts)
```

## 4. Scaling

```python
for n in range(1, 21):
    amplitudes = 2**n
    bytes_complex128 = amplitudes * 16
    print(f"{n:2d} qubits -> {amplitudes:>10,d} amplitudes -> ~{bytes_complex128/1024**2:8.2f} MiB")
```

## 5. Basis-state helper

```python
def basis_state_circuit(bitstring):
    # We interpret bitstring in the same left-to-right form Qiskit displays.
    n = len(bitstring)
    qc = QuantumCircuit(n)
    for displayed_position, bit in enumerate(reversed(bitstring)):
        if bit == "1":
            qc.x(displayed_position)
        elif bit != "0":
            raise ValueError("Use only 0 and 1.")
    return qc

test = basis_state_circuit("101")
print(Statevector.from_instruction(test).probabilities_dict())
test.draw("mpl")
```
